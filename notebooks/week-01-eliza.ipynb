## 1 · The whole program

Three ingredients: a pattern table, a pronoun-reflection map, and a fallback list.
That is all ELIZA ever was.

```python
import re
import random

random.seed(1)   # reproducible in class; remove for live improvisation

RULES = [
    (r"\bI am (.*)", ["How long have you been {0}?",
                      "Why do you say you are {0}?"]),
    (r"\bI feel (.*)", ["Do you often feel {0}?"]),
    (r"\bmy (mother|father|family)\b", ["Tell me more about your {0}."]),
    (r"\bbecause\b", ["Is that the real reason?"]),
    (r"\bno\b", ["Why not?"]),
    (r"\b(computer|machine)s?\b", ["Do machines worry you?"]),
]

FALLBACK = ["Please go on.", "What does that suggest to you?",
            "Can you elaborate on that?"]

REFLECT = {"i": "you", "me": "you", "my": "your", "am": "are",
           "you": "I", "your": "my", "are": "am"}


def reflect(text):
    return " ".join(REFLECT.get(w, w) for w in text.lower().split())


def respond(text):
    for pattern, templates in RULES:
        m = re.search(pattern, text, re.IGNORECASE)
        if m:
            groups = [reflect(g) for g in m.groups()]
            return random.choice(templates).format(*groups)
    return random.choice(FALLBACK)


print("ELIZA is", sum(len(t) for _, t in RULES) + len(FALLBACK), "canned responses.")
```

## 2 · It works

Run this. Have the chat feed you inputs. It will feel eerily plausible for about
six exchanges — which is exactly how long Weizenbaum's users needed to form
attachments.

```python
conversation = [
    "I am not feeling great about the exam",
    "I feel anxious most of the time",
    "My mother thinks I should study more",
    "I did badly because I did not sleep",
]

for line in conversation:
    print(f"you  > {line}")
    print(f"ELIZA> {respond(line)}\n")
```

## 3 · Now break it — this is the point of the exercise

Four failures, each a different *kind* of failure. Read the annotations before
running: predicting the failure mode is more valuable than seeing it.

```python
probes = [
    ("I am not feeling great about the exam",
     "WORKS — pattern matched, pronouns reflected"),
    ("Do you think I should drop this course?",
     "NO MODEL OF THE QUESTION — a direct request for advice gets a deflection"),
    ("My mother is a doctor and my father is a lawyer",
     "IGNORED HALF THE SENTENCE — first match wins, the rest is invisible"),
    ("I am my own worst enemy",
     "ACCIDENTAL PROFUNDITY — syntactically perfect, semantically empty"),
]

random.seed(1)
for text, note in probes:
    print(f"you  > {text}")
    print(f"ELIZA> {respond(text)}")
    print(f"        ^ {note}\n")
```

## 4 · Try it yourself

Uncomment and run for an interactive loop. Type `quit` to exit.

In a virtual class it works better to have students post inputs in the chat and
paste them here — the whole room sees the same failure at the same moment.

```python
while True:
    text = input("you  > ")
    if text.strip().lower() in {"quit", "exit", ""}:
        break
    print("ELIZA>", respond(text))
```

## 5 · The discussion

> **ELIZA has no model of anything.** No representation of the user, the topic, or
> the world. It is 6 regular expressions and a dictionary.
>
> Users in 1966 formed emotional attachments to it anyway. **Weizenbaum's own
> secretary asked him to leave the room** so she could talk to it privately.

**What does that tell us about the Turing Test as a measure of intelligence?**

The uncomfortable answer: it tells us the test measures something about the
*observer* as much as the system. Weizenbaum spent the rest of his career arguing
this point.
